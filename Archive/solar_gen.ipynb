```python
import pandas as pd
import numpy as np
from xgboost import XGBRegressor

from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error

import plotly.express as px
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('cleanDataV2.csv')
data=data.rename(columns={'TIMESTAMP':'date'})
data['date']=pd.to_datetime(data['date'])
data.set_index('date', inplace=True)
data.drop('GI_TOTAL_KWH_ENERGIA', axis=1, inplace=True)
data.head()
```

```python
data['month']=pd.DatetimeIndex(data.index).month
data['day']=pd.DatetimeIndex(data.index).day
data['hour']=pd.DatetimeIndex(data.index).hour
```

```python
data.isna().mean()
```

```python
import statsmodels.api as sm


train=data[data.index<'2024-09-01 00:00:00']
test=data[(data.index>='2024-09-01 00:00:00')&(data.index<'2024-10-01 00:00:00')]

x_train, y_train=train.drop('SG_TOTAL_KWH_ENERGIA', axis=1), train['SG_TOTAL_KWH_ENERGIA']
x_test, y_test=test.drop('SG_TOTAL_KWH_ENERGIA', axis=1), test['SG_TOTAL_KWH_ENERGIA']


x=sm.add_constant(x_train)
y=data['SG_TOTAL_KWH_ENERGIA']
# x=data.drop('SG_TOTAL_KWH_ENERGIA', axis=1)
# y=data['SG_TOTAL_KWH_ENERGIA']

model=sm.OLS(y_train, x)
results=model.fit()
results.summary()

```

```python
significant_vars = results.pvalues[results.pvalues < 0.05].index.tolist()
significant_vars
```

```python
significant_vars.remove('const')
significant_vars.append('SG_TOTAL_KWH_ENERGIA')
```

```python
data2=data[significant_vars]
data2.head()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
variables=data2.drop('SG_TOTAL_KWH_ENERGIA', axis=1)
vif=pd.DataFrame()
vif['VIF']=[variance_inflation_factor(variables.values, i) for i in range(variables.shape[1])]
vif['Features']=variables.columns

with pd.option_context('display.float_format', '{:.2f}'.format):
    print(vif.sort_values('VIF', ascending=False))
```

```python
corr_matrix = data2.corr()
sns.set_theme(font_scale=0.5)

high_corr = corr_matrix[abs(corr_matrix) > 0.3]
fig, ax = plt.subplots(figsize=(8,8))
sns.heatmap(high_corr, annot=True, fmt=".2f", linewidths=1, ax=ax)
```

```python
train=data[data.index<'2024-09-01 00:00:00']
test=data[(data.index>='2024-09-01 00:00:00')&(data.index<'2024-10-01 00:00:00')]

#train.dropna(inplace=True)

#test.dropna(subset=test.columns.drop('TOTAL_KWH_ENERGIA'), inplace=True)

x_train, y_train= train.drop('SG_TOTAL_KWH_ENERGIA', axis=1), train['SG_TOTAL_KWH_ENERGIA']
x_test, y_test=test.drop('SG_TOTAL_KWH_ENERGIA', axis=1), test['SG_TOTAL_KWH_ENERGIA']
```

```python
#Define the model

xgb=XGBRegressor(objective='reg:squarederror', learning_rate=0.1, n_estimators=300, subsample=0.3)

xgb.fit(x_train, y_train)

preds=xgb.predict(x_test)

preds=[0 if pred<2 else pred for pred in preds]

mae=mean_absolute_error(y_test,preds)

print(f'MAE:{mae:.2f}')
```
